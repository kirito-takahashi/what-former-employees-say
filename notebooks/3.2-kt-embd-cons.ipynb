```python
import pandas as pd
import numpy as np
import torch
from google.colab import drive, userdata
from sentence_transformers import SentenceTransformer

drive.mount("/content/drive")

HF_TOKEN = userdata.get('HF_TOKEN')

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

# 1. Load Data

```python
train = pd.read_parquet("train.parquet")
val = pd.read_parquet("validation.parquet")
test = pd.read_parquet("test.parquet")
```

```python
# Download model
model = SentenceTransformer("google/embeddinggemma-2", token=HF_TOKEN)
model.max_seq_length = 512
```

# 2. Embed cons texts with (L4 GPU)

```python
def embed_texts(name, df, model, device, prompt_name="Classification"):
    print(f"====={name}=====")
    embd = model.encode(df["cons_clean"].to_list(),
                        prompt_name=prompt_name,
                        device=device,
                        show_progress_bar=True
                        )
    np.save(f"{name}_embd.npy", embd)
    print(f"====={name}.npy was saved=====")
    return embd
```

```python
train_embd = embed_texts("train", train, model, device)
val_embd = embed_texts("validation", val, model, device)
test_embd = embed_texts("test", test, model, device)
```
